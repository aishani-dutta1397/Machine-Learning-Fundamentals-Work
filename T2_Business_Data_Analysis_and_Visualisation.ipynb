```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("2. Finance_Analytics_Dataset.csv")
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
print("Missing Values:\n",df.isnull().sum())
```

```python
print("Duplicates:",df.duplicated().sum())
```

```python
#number of transactions in each category using count plot
sns.countplot(x = "Category", data = df)
plt.show()
```

```python
#number of transactions for each payment mode using count plot
sns.countplot(x = "PaymentMode", data = df)
plt.show()
```

```python
#number of transactions for each transaction type using count plot
sns.countplot(x = "TransactionType", data = df)
plt.show()
```

```python
#debit vs credit amount comparison using bar plot
sns.barplot(x = "TransactionType", y = "Amount", data = df)
plt.show()
```

```python
#spending pattern by customer segment bar plot
sns.barplot(x = "CustomerSegment", y = "Amount", data = df)
plt.show()
```

```python
#payment mode usage distribution pie chart
paymentmode_counts = df['PaymentMode'].value_counts()
plt.pie(paymentmode_counts, labels=paymentmode_counts.index, autopct='%1.1f%%')
plt.title("Payment Mode Usage Distribution")
plt.show()
```

```python
#boxplot of transaction amount by category
sns.boxplot(x = "Category", y = "Amount", data = df)
plt.show()
```
