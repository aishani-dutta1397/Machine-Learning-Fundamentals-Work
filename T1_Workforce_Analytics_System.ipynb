```python
import pandas as pd
```

```python
data = pd.read_csv("1. hr_analytics.xls")
```

```python
df = pd.DataFrame(data)
```

```python
print("Rows and Columns:\n", df.shape)
```

```python
print("Data Types of Each Column:\n", df.dtypes)
```

```python
print("Missing Values:\n",df.isnull().sum())
```

```python
print("Duplicates:",df.duplicated().sum())
```

```python
total_employees = len(df)
print("Total Employees:", total_employees)
```

```python
print("Total Departments:" , df["Department"].nunique())
```

```python
print("Departments:" , df["Department"].unique())
```

```python
print("Average Employee Age:", df["Age"].mean())
```

```python
print("Minimum Employee Age:", df["Age"].min())
```

```python
print("Maximum Employee Age:", df["Age"].max())
```

```python
print("Gender Distribution:\n", df["Gender"].value_counts())
```

```python
print("Employees in Each Department:\n", df["Department"].value_counts())
```

```python
employee_tenure = df[df["TotalWorkingYears"] > 10]
print("Employees with more than 10 years of experience:", len(employee_tenure))
```

```python
employee_tenure_5 = df[df["TotalWorkingYears"] > 5]
print("Employees who have been with the company for more than 5 years:", len(employee_tenure_5))
```

```python
print("Average Monthly Income in Each Department:\n", df.groupby("Department")["MonthlyIncome"].mean())
```
