```python
import pandas as pd
```

```python
data = pd.read_csv("customer_orders.csv")
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isnull().sum()
```

```python
#1a
df['CustomerName'] = df['CustomerName'].fillna(value = "Unknown")
```

```python
#1b
df['Product'] = df['Product'].fillna(value = "Unknown Product")
```

```python
#1c
df.dropna(subset=['Quantity'], inplace=True)
```

```python
#1d
df['PricePerUnit'] = df['PricePerUnit'].fillna(value = df['PricePerUnit'].mean())
```

```python
#2a
df['Quantity'] = df['Quantity'].abs()
```

```python
#3a
df = df.drop_duplicates()
```

```python
#4a
df['Quantity'] = df['Quantity'].astype('Int64')
```

```python
#5a
df['TotalPrice'] = df['Quantity'] * df['PricePerUnit']
```
